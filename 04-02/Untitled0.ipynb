```python
import requests

url = 'http://openapi.seoul.go.kr:8088//인증키/json/SeoulPublicLibraryInfo/1/100/'

response = requests.get(url)
print(response.content)

```

```python
data = response.json()
data
```

```python
import pandas as pd

temp=[]

for d in data['SeoulPublicLibraryInfo']['row']:
  temp.append(d)

uni = pd.DataFrame(temp)
uni.head()
```

```python
import folium
anMap = folium.Map(location=[37.51384422, 127.02710693],
                   zoom_start=11,
                   zoom_control=False,
                   control_scale=True)
anMap
```

```python
from folium.plugins import MarkerCluster
mc = MarkerCluster().add_to(anMap)

for n, row in uni.iterrows():
  folium.Marker(location=[row['XCNTS'], row['YDNTS']], popup = folium.Popup(row['LBRRY_NAME'], max_width=200)).add_to(mc)

anMap
```
